### Modern C11 Types (`_Generic`)
Demonstrates type-generic programming using C11's built-in `_Generic` selection.

```c
//%cflags:-std=c11 -Wall
#include <stdio.h>

#define print_type(x) _Generic((x), \
    int: "Type is an Integer", \
    double: "Type is a Double", \
    default: "Type is Other" \
)

int main(void) {
    int a = 42;
    double b = 3.14;
    
    printf("Variable a: %s\n", print_type(a));
    printf("Variable b: %s\n", print_type(b));
    
    return 0;
}
```

### Anonymous Structures & Unions (C11 Feature)
Tests C11 anonymous structures and unions nested directly inside other structs.

```c
//%cflags:-std=c11 -Wall
#include <stdio.h>

struct Vector {
    union {
        struct { float x, y, z; };
        float v[3];
    };
};

int main(void) {
    struct Vector vec;
    vec.x = 1.0f;
    vec.y = 2.5f;
    vec.z = 3.7f;
    
    printf("Vector coordinates: %.1f, %.1f, %.1f\n", vec.v[0], vec.v[1], vec.v[2]);
    return 0;
}
```

# Dynamic array size

```c
#include <stdio.h>
#include <stdlib.h>

struct nlist {
    int x;
    struct nlist *next;
    struct nlist *last;
};

/* Create a new node */
struct nlist* nlist_new(int x) {
    struct nlist *node = (struct nlist *)malloc(sizeof(struct nlist));
    node->x = x;
    node->next = NULL;
    node->last = node;
    return node;
}

/* Add a new element to the list */
void nlist_add(struct nlist *self, int x) {
    struct nlist *node = nlist_new(x);
    self->last->next = node;
    self->last = node;
}

/* Print all elements in the list */
void print_elements(const struct nlist *self) {
    const struct nlist *current = self;
    printf("Elements: ");
    while (current != NULL) {
        printf("%d ", current->x);
        current = current->next;
    }
    printf("\n");
}

int main(void) {
    int i;
    int sum = 0;
    
    /* Initialize the list with the first element (1) */
    struct nlist *head = nlist_new(1);
    sum += 1;

    /* Populate the list with 2 through 100 and calculate the sum */
    for (i = 2; i <= 100; i++) {
        nlist_add(head, i);
        sum += i;
    }

    /* Print all elements using our custom function */
    print_elements(head);

    /* Output the final sum */
    printf("Sum of 1 through 100 is: %d\n", sum);

    /* Clean up allocated memory */
    {
        struct nlist *curr = head;
        while (curr != NULL) {
            struct nlist *temp = curr;
            curr = curr->next;
            free(temp);
        }
    }

    return 0;
}
```

# Static Array Size

```c
#include <stdio.h>

/* Custom function to print array elements */
void print_elements(const int *arr, int size) {
    int i;
    printf("Elements: ");
    for (i = 0; i < size; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
}

int main(void) {
    int numbers[100];
    int i;
    int sum = 0;

    /* Populate the array and calculate the sum */
    for (i = 0; i < 100; i++) {
        numbers[i] = i + 1;
        sum += numbers[i];
    }

    /* Print all elements using our custom function */
    print_elements(numbers, 100);

    /* Output the final sum */
    printf("Sum of 1 through 100 is: %d\n", sum);

    return 0;
}
```
